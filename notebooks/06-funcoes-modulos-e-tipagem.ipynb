## Definindo e retornando valores

Funções agrupam uma responsabilidade. `return` devolve o resultado; sem ele, a função retorna `None`.


```python
def calcular_media(valores):
    '''Retorna a média aritmética de uma sequência não vazia.'''
    return sum(valores) / len(valores)


notas = [8.0, 9.5, 7.0]
print(calcular_media(notas))
print(calcular_media.__doc__)
```

## Parâmetros posicionais, nomeados e padrões

Argumentos nomeados deixam a chamada explícita. Valores padrão são avaliados uma vez; evite listas e dicionários mutáveis como padrão.


```python
def apresentar(nome, saudacao="Olá", pontuacao="!"):
    return f"{saudacao}, {nome}{pontuacao}"


print(apresentar("Lia"))
print(apresentar("Caio", saudacao="Bem-vindo", pontuacao="."))
```

## Quantidade variável de argumentos

`*args` reúne argumentos posicionais em uma tupla e `**kwargs` reúne argumentos nomeados em um dicionário.


```python
def somar(*valores):
    return sum(valores)


def criar_perfil(nome, **detalhes):
    return {"nome": nome, **detalhes}


print(somar(2, 4, 6, 8))
print(criar_perfil("Ana", cidade="Salvador", linguagem="Python"))
```

## Escopo

Nomes definidos em uma função são locais. Funções podem ler nomes externos, mas receber dependências por parâmetro geralmente torna o código mais previsível.


```python
taxa_padrao = 0.10

def calcular_total(valor, taxa=taxa_padrao):
    subtotal = valor * (1 + taxa)
    return round(subtotal, 2)


print(calcular_total(100))
```

## Funções como objetos e `lambda`

Funções podem ser atribuídas a variáveis e passadas para outras funções. `lambda` é adequada para expressões pequenas.


```python
produtos = [
    {"nome": "Mouse", "preco": 90},
    {"nome": "Monitor", "preco": 1200},
    {"nome": "Teclado", "preco": 250},
]

por_preco = sorted(produtos, key=lambda produto: produto["preco"])
print(por_preco)
```

## Anotações de tipo

Type hints documentam valores esperados e ajudam editores e verificadores estáticos. Python continua dinamicamente tipado em tempo de execução.


```python
def totalizar(precos: list[float], desconto: float = 0.0) -> float:
    total = sum(precos)
    return round(total * (1 - desconto), 2)


print(totalizar([10.0, 20.0, 30.0], desconto=0.10))
print(totalizar.__annotations__)
```

## Módulos e biblioteca padrão

Um módulo é um arquivo Python. `import` permite reutilizar definições. A biblioteca padrão inclui ferramentas para matemática, datas, caminhos, estatística e muito mais.


```python
from datetime import date, timedelta
from statistics import mean, median

hoje = date(2026, 9, 11)
prazo = hoje + timedelta(days=15)

print("Prazo:", prazo.isoformat())
print("Média e mediana:", mean([2, 5, 9]), median([2, 5, 9]))
```

## Recursão

Uma função recursiva chama a si mesma e precisa de um caso-base. Para repetições simples, laços costumam ser mais claros.


```python
def fatorial(numero: int) -> int:
    if numero < 0:
        raise ValueError("O número deve ser não negativo")
    if numero in (0, 1):
        return 1
    return numero * fatorial(numero - 1)


print(fatorial(5))
```

## Pratique

Crie uma função que receba preços e um percentual de desconto opcional, recusando descontos fora do intervalo de 0 a 1.


```python
# Solução sugerida
def aplicar_desconto(precos: list[float], desconto: float = 0.0) -> list[float]:
    if not 0 <= desconto <= 1:
        raise ValueError("O desconto deve estar entre 0 e 1")
    return [round(preco * (1 - desconto), 2) for preco in precos]


print(aplicar_desconto([100, 80, 50], 0.15))
```

## Resumo

Funções pequenas, documentadas e com entradas explícitas são mais fáceis de testar e reutilizar. Módulos distribuem essas funções em arquivos coerentes.
