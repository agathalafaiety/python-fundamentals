# 05 — Laços, iteração e compreensões

**Objetivos:** percorrer coleções com `for`, usar ferramentas de iteração e construir coleções de forma expressiva.


## `for` e objetos iteráveis

`for` percorre itens de qualquer objeto iterável, como listas, strings, dicionários e arquivos.


```python
linguagens = ["Python", "JavaScript", "SQL"]

for linguagem in linguagens:
    print(linguagem.upper())
```

## `range`, `enumerate` e `zip`

- `range` produz sequências numéricas;
- `enumerate` fornece índice e valor;
- `zip` percorre coleções em paralelo.


```python
nomes = ["Ana", "Bia", "Caio"]
notas = [8.5, 9.0, 7.5]

for posicao, (nome, nota) in enumerate(zip(nomes, notas), start=1):
    print(f"{posicao}. {nome}: {nota}")

print("Quadrados:", [numero ** 2 for numero in range(1, 6)])
```

## Iterando dicionários

`items()` devolve pares de chave e valor; `keys()` e `values()` fornecem cada parte separadamente.


```python
estoque = {"caderno": 12, "caneta": 30, "mochila": 5}

for produto, quantidade in estoque.items():
    print(f"{produto.title():10} | {quantidade:2} unidades")
```

## `break`, `continue` e `else`

`break` encerra a busca, `continue` ignora o restante da iteração e o `else` de um laço roda apenas quando não houve `break`.


```python
numeros = [3, 7, 11, 18, 21]

for numero in numeros:
    if numero % 2 == 0:
        print("Primeiro par:", numero)
        break
else:
    print("Nenhum número par encontrado")
```

## Compreensões

Compreensões criam listas, dicionários e conjuntos a partir de um iterável. Use-as quando a expressão permanecer curta e clara.


```python
numeros = range(1, 11)

pares_ao_quadrado = [n ** 2 for n in numeros if n % 2 == 0]
mapa_quadrados = {n: n ** 2 for n in range(1, 6)}
iniciais = {nome[0] for nome in ["Ana", "Alice", "Bruno", "Bianca"]}

print(pares_ao_quadrado)
print(mapa_quadrados)
print(iniciais)
```

## Geradores

Uma expressão geradora usa parênteses e produz os valores sob demanda, economizando memória em grandes sequências.


```python
quadrados = (n ** 2 for n in range(1, 1_000_001))
primeiros = [next(quadrados) for _ in range(5)]
print(primeiros)
```

## Funções úteis em iterações

`sum`, `min`, `max`, `any`, `all`, `sorted` e `reversed` resolvem muitos problemas sem laços manuais.


```python
notas = [8.0, 7.5, 9.5, 6.0]

print("Média:", sum(notas) / len(notas))
print("Todas aprovadas?", all(nota >= 6 for nota in notas))
print("Alguma nota máxima?", any(nota == 10 for nota in notas))
```

## Pratique

A partir de uma lista de palavras, crie um dicionário contendo apenas as palavras com cinco ou mais letras e seus respectivos tamanhos.


```python
# Solução sugerida
palavras = ["sol", "python", "dados", "web", "função"]
tamanhos = {palavra: len(palavra) for palavra in palavras if len(palavra) >= 5}
print(tamanhos)
```

## Resumo

Ferramentas como `enumerate`, `zip` e compreensões tornam iterações mais declarativas. Priorize legibilidade e prefira um laço tradicional quando a regra for complexa.
