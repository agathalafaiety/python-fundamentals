## Saída, comentários e variáveis

`print()` envia valores para a saída. Comentários começam com `#`. Variáveis são nomes que apontam para objetos; em Python não é necessário declarar o tipo antes da atribuição. Use nomes descritivos em `snake_case`.


```python
# Minha primeira instrução em Python
curso = "Python Fundamentals"
aulas = 10
concluido = False

print(curso)
print("Quantidade de aulas:", aulas)
print("Concluído?", concluido)
```

## Tipos básicos

- `int`: números inteiros;
- `float`: números com parte decimal;
- `str`: texto;
- `bool`: `True` ou `False`;
- `None`: ausência intencional de valor.

A função `type()` mostra o tipo de um objeto.


```python
exemplos = [42, 3.14, "Olá", True, None]

for valor in exemplos:
    print(f"{valor!r:8} -> {type(valor).__name__}")
```

## Conversão de tipos

Use `int()`, `float()`, `str()` e `bool()` para conversões explícitas. Nem toda conversão é possível: por exemplo, `int("Python")` produziria um `ValueError`.


```python
idade_texto = "25"
idade = int(idade_texto)
preco = float("19.90")

print(idade + 1)
print(preco * 2)
print(str(idade) + " anos")
```

## Entrada de dados

`input()` sempre devolve uma string. Em um programa interativo poderíamos usar `nome = input("Nome: ")`. Para manter este notebook totalmente automático, simularemos a resposta do usuário.


```python
entrada_simulada = "Agatha"
nome = entrada_simulada.strip()
print(f"Olá, {nome}!")
```

## Operadores aritméticos

Python oferece soma (`+`), subtração (`-`), multiplicação (`*`), divisão (`/`), divisão inteira (`//`), resto (`%`) e potência (`**`). Parênteses deixam a precedência explícita.


```python
a, b = 17, 5

resultados = {
    "soma": a + b,
    "subtração": a - b,
    "multiplicação": a * b,
    "divisão": a / b,
    "divisão inteira": a // b,
    "resto": a % b,
    "potência": a ** 2,
}
resultados
```

## Atribuição e constantes por convenção

Operadores como `+=` e `*=` atualizam uma variável. Python não impede a alteração de constantes, mas nomes em `MAIUSCULAS` comunicam que o valor não deve mudar.


```python
PI = 3.14159
pontos = 10
pontos += 5
pontos *= 2

raio = 3
area = PI * raio ** 2
print(f"Pontos: {pontos}")
print(f"Área: {area:.2f}")
```

## Pratique

Converta uma temperatura de Celsius para Fahrenheit usando a fórmula `F = C × 9/5 + 32`.


```python
# Solução sugerida
celsius = 24
fahrenheit = celsius * 9 / 5 + 32
print(f"{celsius} °C equivalem a {fahrenheit:.1f} °F")
```
