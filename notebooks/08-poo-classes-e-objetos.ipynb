## Classe, objeto, atributo e método

Uma classe define um tipo. Cada objeto é uma instância com seus próprios atributos. Métodos descrevem comportamentos e recebem a própria instância em `self`.


```python
class ContaBancaria:
    banco = "Banco Python"  # atributo de classe

    def __init__(self, titular: str, saldo: float = 0.0):
        self.titular = titular
        self.saldo = float(saldo)

    def depositar(self, valor: float) -> None:
        if valor <= 0:
            raise ValueError("O depósito deve ser positivo")
        self.saldo += valor

    def exibir_saldo(self) -> str:
        return f"{self.titular}: R$ {self.saldo:.2f}"


conta = ContaBancaria("Ana", 100)
conta.depositar(50)
print(conta.exibir_saldo())
print(conta.banco)
```

## Encapsulamento e propriedades

Python usa convenções: um nome iniciado por `_` é considerado interno. `property` permite controlar a leitura e a alteração mantendo uma interface simples.


```python
class Produto:
    def __init__(self, nome: str, preco: float):
        self.nome = nome
        self.preco = preco

    @property
    def preco(self) -> float:
        return self._preco

    @preco.setter
    def preco(self, valor: float) -> None:
        if valor < 0:
            raise ValueError("O preço não pode ser negativo")
        self._preco = float(valor)


produto = Produto("Teclado", 250)
produto.preco = 230
print(produto.preco)
```

## Métodos de classe e estáticos

`classmethod` recebe a classe em `cls` e pode atuar como construtor alternativo. `staticmethod` agrupa uma função relacionada à classe que não precisa da instância.


```python
from datetime import date

class Pessoa:
    def __init__(self, nome: str, ano_nascimento: int):
        self.nome = nome
        self.ano_nascimento = ano_nascimento

    @classmethod
    def a_partir_da_idade(cls, nome: str, idade: int):
        return cls(nome, date.today().year - idade)

    @staticmethod
    def maior_de_idade(idade: int) -> bool:
        return idade >= 18


pessoa = Pessoa.a_partir_da_idade("Caio", 25)
print(pessoa.nome, pessoa.ano_nascimento)
print(Pessoa.maior_de_idade(25))
```

## Representação e comparação

Métodos especiais integram objetos à linguagem. `__repr__` cria uma representação útil para desenvolvimento e `__eq__` define igualdade por valor.


```python
class Ponto:
    def __init__(self, x: int, y: int):
        self.x = x
        self.y = y

    def __repr__(self) -> str:
        return f"Ponto(x={self.x}, y={self.y})"

    def __eq__(self, outro) -> bool:
        if not isinstance(outro, Ponto):
            return NotImplemented
        return (self.x, self.y) == (outro.x, outro.y)


print(Ponto(2, 3))
print(Ponto(2, 3) == Ponto(2, 3))
```

## Dataclasses

`@dataclass` gera inicialização, representação e igualdade para classes centradas em dados. Métodos continuam podendo ser adicionados normalmente.


```python
from dataclasses import dataclass, field

@dataclass
class Pedido:
    numero: int
    itens: list[float] = field(default_factory=list)

    @property
    def total(self) -> float:
        return round(sum(self.itens), 2)


pedido = Pedido(101, [19.9, 30.0])
print(pedido)
print("Total:", pedido.total)
```

## Responsabilidades

Uma classe deve ter um propósito claro. Prefira validar o estado na entrada e colocar junto dos dados os comportamentos que dependem deles.


```python
@dataclass
class Retangulo:
    largura: float
    altura: float

    def __post_init__(self):
        if self.largura <= 0 or self.altura <= 0:
            raise ValueError("As dimensões devem ser positivas")

    def area(self) -> float:
        return self.largura * self.altura

    def perimetro(self) -> float:
        return 2 * (self.largura + self.altura)


retangulo = Retangulo(5, 3)
print(retangulo.area(), retangulo.perimetro())
```

## Pratique

Crie uma classe `Carrinho` com métodos para adicionar preços e calcular o total.


```python
# Solução sugerida
class Carrinho:
    def __init__(self):
        self._precos = []

    def adicionar(self, preco: float) -> None:
        if preco < 0:
            raise ValueError("Preço inválido")
        self._precos.append(float(preco))

    @property
    def total(self) -> float:
        return round(sum(self._precos), 2)


carrinho = Carrinho()
carrinho.adicionar(29.90)
carrinho.adicionar(15.50)
print(carrinho.total)
```

## Resumo

Classes reúnem estado e comportamento. Construtores estabelecem um estado válido, propriedades protegem regras e dataclasses reduzem código repetitivo.
