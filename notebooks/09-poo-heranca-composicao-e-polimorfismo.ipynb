## Herança e `super()`

Uma subclasse herda atributos e métodos. `super()` delega parte da inicialização ou do comportamento para a classe-base.


```python
class Funcionario:
    def __init__(self, nome: str, salario: float):
        self.nome = nome
        self.salario = salario

    def calcular_bonus(self) -> float:
        return self.salario * 0.05


class Gerente(Funcionario):
    def __init__(self, nome: str, salario: float, equipe: int):
        super().__init__(nome, salario)
        self.equipe = equipe

    def calcular_bonus(self) -> float:
        return super().calcular_bonus() + self.equipe * 100


gerente = Gerente("Ana", 8000, 5)
print(gerente.nome, gerente.calcular_bonus())
```

## Polimorfismo

Objetos diferentes podem responder ao mesmo método. Quem usa os objetos depende do comportamento, e não necessariamente da classe concreta.


```python
class Email:
    def enviar(self, mensagem: str) -> str:
        return f"E-mail enviado: {mensagem}"


class SMS:
    def enviar(self, mensagem: str) -> str:
        return f"SMS enviado: {mensagem}"


def notificar(canal, mensagem: str) -> None:
    print(canal.enviar(mensagem))


for canal in [Email(), SMS()]:
    notificar(canal, "Pedido confirmado")
```

## Classes abstratas

Uma classe abstrata define um contrato que as subclasses concretas devem implementar.


```python
from abc import ABC, abstractmethod
from math import pi

class Forma(ABC):
    @abstractmethod
    def area(self) -> float:
        raise NotImplementedError


class Circulo(Forma):
    def __init__(self, raio: float):
        self.raio = raio

    def area(self) -> float:
        return pi * self.raio ** 2


print(f"Área: {Circulo(2).area():.2f}")
```

## Composição

Na composição, um objeto recebe ou cria outros objetos para realizar seu trabalho. É uma relação de “tem um”, geralmente mais flexível que “é um”.


```python
from dataclasses import dataclass

@dataclass
class Motor:
    potencia: int

    def ligar(self) -> str:
        return f"Motor de {self.potencia} cv ligado"


class Carro:
    def __init__(self, modelo: str, motor: Motor):
        self.modelo = modelo
        self.motor = motor

    def iniciar(self) -> str:
        return f"{self.modelo}: {self.motor.ligar()}"


carro = Carro("Modelo P", Motor(150))
print(carro.iniciar())
```

## Protocolos informais e duck typing

Em Python, se um objeto oferece os métodos necessários, ele pode ser utilizado independentemente da sua árvore de herança: “se anda como um pato...”.


```python
class ArquivoEmMemoria:
    def __init__(self):
        self.conteudo = ""

    def write(self, texto: str) -> int:
        self.conteudo += texto
        return len(texto)


def registrar(destino, mensagem: str) -> None:
    destino.write(mensagem + "\n")


destino = ArquivoEmMemoria()
registrar(destino, "Aplicação iniciada")
print(destino.conteudo)
```

## Métodos especiais de coleção

`__len__`, `__iter__` e `__contains__` permitem que objetos funcionem com `len`, `for` e `in`.


```python
class Playlist:
    def __init__(self, musicas=None):
        self._musicas = list(musicas or [])

    def __len__(self):
        return len(self._musicas)

    def __iter__(self):
        return iter(self._musicas)

    def __contains__(self, musica):
        return musica in self._musicas


playlist = Playlist(["Música A", "Música B"])
print(len(playlist), "Música A" in playlist)
print(list(playlist))
```

## Pratique

Modele duas formas de pagamento com o mesmo método `pagar` e processe ambas por uma única função.


```python
# Solução sugerida
class Cartao:
    def pagar(self, valor: float) -> str:
        return f"R$ {valor:.2f} pagos no cartão"


class Pix:
    def pagar(self, valor: float) -> str:
        return f"R$ {valor:.2f} pagos via Pix"


def finalizar_pagamento(meio, valor: float) -> None:
    print(meio.pagar(valor))


finalizar_pagamento(Cartao(), 120)
finalizar_pagamento(Pix(), 75.5)
```

## Resumo

Use herança para especializar um contrato estável e composição para combinar responsabilidades. Polimorfismo reduz acoplamento entre quem solicita e quem executa um comportamento.
