# 03 — Coleções: listas, tuplas, conjuntos e dicionários

**Objetivos:** armazenar vários valores, escolher a estrutura adequada e compreender mutabilidade.


## Listas

Listas são ordenadas, mutáveis e aceitam valores repetidos. Índices começam em zero; índices negativos contam a partir do fim.


```python
frutas = ["maçã", "banana", "uva"]
frutas.append("manga")
frutas.insert(1, "pera")

print(frutas[0], frutas[-1])
print(frutas[1:4])
print("Quantidade:", len(frutas))
```

Métodos comuns: `append`, `extend`, `insert`, `remove`, `pop`, `sort`, `reverse`, `index` e `count`. `sorted()` devolve uma nova lista sem alterar a original.


```python
numeros = [8, 3, 5, 3, 1]
ordenados = sorted(numeros)
ultimo = numeros.pop()

print("Original após pop:", numeros)
print("Ordenados:", ordenados)
print("Valor removido:", ultimo)
print("Quantidade de 3:", numeros.count(3))
```

## Cópia e mutabilidade

`b = a` cria outro nome para a mesma lista. Use `a.copy()` ou um fatiamento para uma cópia rasa.


```python
original = [1, 2, 3]
copia = original.copy()
copia.append(4)

print("Original:", original)
print("Cópia:", copia)
```

## Tuplas e desempacotamento

Tuplas são ordenadas e imutáveis. Funcionam bem para registros pequenos e retornos múltiplos.


```python
coordenada = (-23.55, -46.63)
latitude, longitude = coordenada
primeiro, *meio, ultimo = [10, 20, 30, 40, 50]

print(latitude, longitude)
print(primeiro, meio, ultimo)
```

## Conjuntos

Conjuntos (`set`) guardam elementos únicos e oferecem operações matemáticas de união, interseção e diferença. Eles não devem ser usados quando a ordem for essencial.


```python
backend = {"Python", "SQL", "Docker"}
frontend = {"JavaScript", "HTML", "Docker"}

print("União:", backend | frontend)
print("Interseção:", backend & frontend)
print("Só no backend:", backend - frontend)
```

## Dicionários

Dicionários associam chaves únicas a valores. São ideais para representar objetos com campos nomeados.


```python
pessoa = {
    "nome": "Ana",
    "idade": 28,
    "habilidades": ["Python", "SQL"],
}
pessoa["cidade"] = "Recife"
pessoa["idade"] += 1

print(pessoa["nome"])
print(pessoa.get("telefone", "não informado"))
print(pessoa)
```

## Estruturas aninhadas

Coleções podem conter outras coleções. Mantenha a estrutura simples e use nomes intermediários quando o acesso ficar difícil de ler.


```python
alunos = [
    {"nome": "Lia", "notas": [8.0, 9.0]},
    {"nome": "Caio", "notas": [7.5, 6.5]},
]

primeira_nota_de_caio = alunos[1]["notas"][0]
print(primeira_nota_de_caio)
```

## Pratique

Remova duplicatas de uma lista de tecnologias e conte a frequência de cada item.


```python
# Solução sugerida
tecnologias = ["Python", "SQL", "Python", "Git", "SQL", "Python"]
unicas = sorted(set(tecnologias))
frequencias = {item: tecnologias.count(item) for item in unicas}

print(unicas)
print(frequencias)
```

## Quando usar cada coleção?

Use lista para sequência mutável, tupla para sequência fixa, conjunto para unicidade e operações de pertencimento, e dicionário para pares de chave e valor.
