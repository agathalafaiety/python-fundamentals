**Objetivos:** comparar valores, combinar condições, tomar decisões com `if` e repetir instruções com `while`.


## Comparações e lógica booleana

Comparações devolvem `True` ou `False`: `==`, `!=`, `<`, `<=`, `>` e `>=`. Os operadores `and`, `or` e `not` combinam expressões booleanas.


```python
idade = 20
possui_documento = True

maior_de_idade = idade >= 18
pode_entrar = maior_de_idade and possui_documento

print("Maior de idade:", maior_de_idade)
print("Entrada permitida:", pode_entrar)
print("Entrada negada:", not pode_entrar)
```

## Pertencimento e identidade

`in` verifica se um valor pertence a uma coleção. `is` verifica identidade e é usado principalmente com `None`; para comparar valores, use `==`.


```python
linguagens = ["Python", "JavaScript", "SQL"]
linguagem = "Python"
resposta = None

print(linguagem in linguagens)
print("Java" not in linguagens)
print(resposta is None)
```

## Condicionais

`if`, `elif` e `else` selecionam um caminho. A indentação define os blocos e deve ser consistente.


```python
nota = 8.2

if nota >= 9:
    conceito = "A"
elif nota >= 7:
    conceito = "B"
elif nota >= 5:
    conceito = "C"
else:
    conceito = "D"

print(f"Conceito: {conceito}")
```

## Valores truthy e falsy

Valores vazios (`""`, `[]`, `{}`), zero e `None` são considerados falsos. Isso permite escrever verificações diretas, como `if nome:`.


```python
carrinho = []

if carrinho:
    print("Há produtos no carrinho")
else:
    print("O carrinho está vazio")

status = "com itens" if carrinho else "vazio"
print(status)
```

## Correspondência de padrões com `match`

A partir do Python 3.10, `match` é útil quando um valor pode assumir várias formas conhecidas.


```python
codigo_status = 404

match codigo_status:
    case 200:
        mensagem = "Sucesso"
    case 400 | 404:
        mensagem = "Erro do cliente"
    case 500:
        mensagem = "Erro do servidor"
    case _:
        mensagem = "Status desconhecido"

print(mensagem)
```

## Repetição com `while`

`while` repete enquanto a condição for verdadeira. Atualize a condição para evitar loops infinitos. `break` encerra o laço e `continue` avança para a próxima repetição.


```python
contador = 0
pares = []

while contador < 10:
    contador += 1
    if contador % 2 != 0:
        continue
    pares.append(contador)
    if len(pares) == 3:
        break

print(pares)
```

## Pratique

Classifique uma compra: frete grátis a partir de R$ 200, frete reduzido a partir de R$ 100 e frete normal nos demais casos.


```python
# Solução sugerida
total = 145.90

if total >= 200:
    frete = "grátis"
elif total >= 100:
    frete = "reduzido"
else:
    frete = "normal"

print(f"Frete {frete}")
```

## Resumo

Comparações produzem booleanos, condicionais escolhem caminhos e `while` controla repetições baseadas em condições.
