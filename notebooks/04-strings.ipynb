# 04 — Strings e processamento de texto

**Objetivos:** acessar, transformar, validar, formatar e pesquisar textos.


## Strings são sequências imutáveis

É possível acessar caracteres e fatiar uma string, mas não alterar uma posição diretamente. Métodos devolvem novas strings.


```python
linguagem = "Python"

print(linguagem[0])
print(linguagem[-1])
print(linguagem[1:4])
print(linguagem[::-1])
```

## Limpeza e transformação

Métodos frequentes incluem `strip`, `lower`, `upper`, `title`, `replace`, `split` e `join`.


```python
texto = "  python é uma linguagem produtiva  "
limpo = texto.strip()
palavras = limpo.split()

print(limpo.capitalize())
print(limpo.replace("produtiva", "versátil"))
print("-".join(palavras))
```

## Busca e validação

Use `in`, `startswith`, `endswith`, `find` e `count` para pesquisas. Métodos como `isdigit`, `isalpha` e `isalnum` ajudam em validações simples.


```python
arquivo = "relatorio_2026.csv"
codigo = "A123"

print(arquivo.endswith(".csv"))
print("2026" in arquivo)
print(codigo.isalnum())
print(codigo[:1].isalpha() and codigo[1:].isdigit())
```

## Formatação

F-strings combinam variáveis e especificadores de formatação de maneira legível.


```python
produto = "Teclado"
preco = 249.9
quantidade = 3

print(f"{produto}: R$ {preco:.2f}")
print(f"Subtotal: R$ {preco * quantidade:,.2f}")
print(f"Código: {quantidade:04d}")
```

## Caracteres especiais e strings multilinha

`\n` representa nova linha e `\t`, tabulação. Prefixos `r` criam strings brutas, úteis para caminhos e expressões regulares.


```python
mensagem = "Nome:\tAgatha\nCurso:\tPython"
caminho_windows = r"C:\\estudos\\python"

print(mensagem)
print(caminho_windows)
```

## Unicode

Strings Python trabalham com Unicode. `casefold()` é mais apropriado do que `lower()` para comparações sem distinção de maiúsculas em diferentes idiomas.


```python
valor_a = "PYTHON"
valor_b = "python"
print(valor_a.casefold() == valor_b.casefold())
```

## Expressões regulares

O módulo `re` identifica padrões mais complexos. Para regras simples, prefira os métodos nativos de string.


```python
import re

texto = "Pedidos: PED-102, PED-305 e PED-999"
codigos = re.findall(r"PED-\d{3}", texto)
print(codigos)
```

## Pratique

Normalize um nome completo removendo espaços extras e colocando cada palavra em formato de título.


```python
# Solução sugerida
nome_digitado = "  maria   das   graças  "
nome_normalizado = " ".join(nome_digitado.split()).title()
print(nome_normalizado)
```
