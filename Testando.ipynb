```python
lista_num = [-10, 2, -6, 4.2, -8, 3.14, -6, 0]
```

```python
lista_num_pos = []
lista_num_neg = []
lista_num_neutro = []

for elemento in lista_num:
  if elemento < 0:
    lista_num_neg.append(elemento)
  elif elemento > 0:
    lista_num_pos.append(elemento)
  else:
    lista_num_neutro.append(elemento)

print(lista_num_pos)
print(lista_num_neg)
print(lista_num_neutro)
```
